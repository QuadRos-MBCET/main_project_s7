```python
# Step 1: Initialize Environment & Memory Monitoring
from ai.config import DEVICE, IS_COLAB
from ai.memory_manager import MemoryManager
from ai.pipeline import run_safead_inference
import json

MemoryManager.print_resource_status()
```

```python
# Step 2: Single Advertisement Inference Demo
# Define target ad path
sample_ad_path = "test_ad.jpg"

# Create sample ad if not present
import os, cv2, numpy as np
if not os.path.exists(sample_ad_path):
    img = np.ones((400, 600, 3), dtype=np.uint8) * 230
    cv2.putText(img, "MEGA CASINO JACKPOT", (50, 200), cv2.FONT_HERSHEY_SIMPLEX, 1, (0, 0, 255), 2)
    cv2.imwrite(sample_ad_path, img)

result = run_safead_inference(
    file_path=sample_ad_path,
    title="Mega Jackpot Casino",
    caption="Win cash fast! Satta khelne ke liye link click karein."
)

print("==========================================================")
print("           STANDARDIZED SAFEAD INFERENCE OUTPUT           ")
print("==========================================================")
print(json.dumps(result, indent=2))
```

```python
# Step 3: Batch Advertisement Processing (Memory-Efficient Sequential Loop)
batch_test_ads = [
    {"path": sample_ad_path, "title": "Online Python Coding Course", "caption": "Learn programming step by step for beginners.", "gt": "SAFE_FOR_ALL"},
    {"path": sample_ad_path, "title": "Weekend Brews Pub", "caption": "Join us for craft draft beer and whiskey tonight.", "gt": "AGE_18_PLUS"},
    {"path": sample_ad_path, "title": "Grand Vegas Casino Satta", "caption": "Earn cash fast! Bet on jackpot slots online.", "gt": "UNSAFE_FOR_ALL"}
]

print("Running Batch Processing...")
for idx, item in enumerate(batch_test_ads):
    res = run_safead_inference(item["path"], item["title"], item["caption"])
    print(f"Ad #{idx+1}: [{item['title']}] -> Classification: {res.get('classification')} | Action: {res.get('action')}")
    # Clear memory between items
    MemoryManager.clear_gpu_memory()
```
