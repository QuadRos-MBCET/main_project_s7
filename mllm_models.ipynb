# Open-Source Multimodal Large Language Models (MLLMs) for SAFE-VISION

This notebook identifies and imports the open-source Multimodal Large Language Models (MLLMs) and Vision-Language Models (VLMs) referenced in the project's research papers. These models are essential for tasks like video moderation, image analysis, and explainable safety guardrails.

### Models Covered:
1. **Qwen2-VL** (General VLM - State of the art)
2. **InternVL2** (General VLM - High-performance visual reasoning)
3. **LLaVA-NeXT-Video** (Video VLM - Specialized in temporal frame-based inference)
4. **Llama-Guard-3-11B-Vision** (Safety Guardrail VLM - Specifically tuned for classifying safe vs. unsafe content)
5. **MiniCPM-V-2.6** (Edge VLM - Extremely efficient VLM with video support)
6. **InstructBLIP / BLIP-2** (Baseline VLMs)

> **Note:** For Kuaishou's self-developed **YuanQi** VLM, since it is a proprietary model used in their internal pipelines, we use equivalent high-performance open-source models like **Qwen2-VL** or **InternVL2** for implementation.

## 1. Qwen2-VL
**Qwen2-VL** is the current state-of-the-art open-source MLLM developed by Alibaba, showing strong performance on video understanding, document parsing, and visual reasoning.

```python
from transformers import Qwen2VLForConditionalGeneration, AutoProcessor

def load_qwen2_vl(model_id="Qwen/Qwen2-VL-7B-Instruct"):
    print(f"Loading {model_id}...")
    model = Qwen2VLForConditionalGeneration.from_pretrained(
        model_id,
        torch_dtype=torch.bfloat16 if torch.cuda.is_available() else torch.float32,
        device_map="auto"
    )
    processor = AutoProcessor.from_pretrained(model_id)
    print("Qwen2-VL successfully loaded.")
    return model, processor

# To load, uncomment the following line:
# model_qwen, processor_qwen = load_qwen2_vl()
```

## 2. InternVL2
**InternVL2** is an extremely capable VLM family by OpenGVLab. It matches proprietary systems like GPT-4o in visual reasoning tasks.

```python
from transformers import AutoModel, AutoTokenizer

def load_internvl2(model_id="OpenGVLab/InternVL2-8B"):
    print(f"Loading {model_id}...")
    # InternVL2 has custom model definitions, requiring trust_remote_code=True
    model = AutoModel.from_pretrained(
        model_id,
        torch_dtype=torch.bfloat16 if torch.cuda.is_available() else torch.float32,
        low_cpu_mem_usage=True,
        trust_remote_code=True,
        device_map="auto"
    ).eval()
    tokenizer = AutoTokenizer.from_pretrained(model_id, trust_remote_code=True, use_fast=False)
    print("InternVL2 successfully loaded.")
    return model, tokenizer

# To load, uncomment the following line:
# model_intern, tokenizer_intern = load_internvl2()
```

## 3. LLaVA-NeXT-Video
**LLaVA-NeXT-Video** is optimized for video reasoning and sequential frames, making it ideal for video content moderation systems.

```python
from transformers import LlavaNextVideoForConditionalGeneration, LlavaNextVideoProcessor

def load_llava_next_video(model_id="llava-hf/LLaVA-NeXT-Video-7B-hf"):
    print(f"Loading {model_id}...")
    model = LlavaNextVideoForConditionalGeneration.from_pretrained(
        model_id,
        torch_dtype=torch.bfloat16 if torch.cuda.is_available() else torch.float32,
        device_map="auto"
    )
    processor = LlavaNextVideoProcessor.from_pretrained(model_id)
    print("LLaVA-NeXT-Video successfully loaded.")
    return model, processor

# To load, uncomment the following line:
# model_llava_video, processor_llava_video = load_llava_next_video()
```

## 4. Llama-Guard-3-11B-Vision
**Llama-Guard-3-11B-Vision** is specifically fine-tuned by Meta to act as a safety classifier/guardrail for multimodal visual prompts and text responses.

```python
from transformers import MllamaForConditionalGeneration, AutoProcessor

def load_llama_guard_3_vision(model_id="meta-llama/Llama-Guard-3-11B-Vision"):
    print(f"Loading {model_id}...")
    # Note: Accessing Llama Guard requires Hugging Face Hub authentication and gated model approval
    model = MllamaForConditionalGeneration.from_pretrained(
        model_id,
        torch_dtype=torch.bfloat16 if torch.cuda.is_available() else torch.float32,
        device_map="auto"
    )
    processor = AutoProcessor.from_pretrained(model_id)
    print("Llama-Guard-3-Vision successfully loaded.")
    return model, processor

# To load, uncomment the following line:
# model_guard, processor_guard = load_llama_guard_3_vision()
```

## 5. MiniCPM-V-2.6
**MiniCPM-V-2.6** is a highly efficient 8B parameter model suitable for edge deployments and machines with lower VRAM, while still supporting video and OCR tasks.

```python
def load_minicpm_v26(model_id="openbmb/MiniCPM-V-2_6"):
    print(f"Loading {model_id}...")
    model = AutoModel.from_pretrained(
        model_id,
        trust_remote_code=True,
        torch_dtype=torch.bfloat16 if torch.cuda.is_available() else torch.float32,
        device_map="auto"
    ).eval()
    tokenizer = AutoTokenizer.from_pretrained(model_id, trust_remote_code=True)
    print("MiniCPM-V-2.6 successfully loaded.")
    return model, tokenizer

# To load, uncomment the following line:
# model_minicpm, tokenizer_minicpm = load_minicpm_v26()
```
